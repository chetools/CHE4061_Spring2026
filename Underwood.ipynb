```python
alpha = np.array([2.5, 2., 1., 0.7])
F=1.
z = np.array([0.25, 0.25, 0.25, 0.25])
q= 0.5
```

```python
lkidx = 1
hkidx = 2

rlkd = 0.95
rhkb = 0.96

wd= F*np.r_[z[:lkidx],rlkd*z[lkidx], (1-rhkb)*z[hkidx], 0*z[hkidx+1:]]
xd = wd/np.sum(wd)
```

```python
def underwood(phi):
    return jnp.sum(alpha*z/(alpha-phi))-(1-q)
```

```python
underwood_vec = jax.jit(jnp.vectorize(underwood, signature='()->()'))
```

```python
fig = make_subplots()
phis = np.linspace(0.9*np.min(alpha),1.1*np.max(alpha),1000)
fig.add_scatter(x=phis, y=underwood_vec(phis))
fig.update_yaxes(range=[-3, 3])
```

```python
phi_root = sp.optimize.root_scalar(underwood,x0=1.5).root
```

```python
min_reflux_ratio = np.sum(alpha*xd/(alpha-phi_root))-1
min_reflux_ratio
```
